# 2 · How a Python program runs

You write `x = 2 + 3` and press run. What actually happens? Python turns your
text into **bytecode** and runs it on a little virtual machine. You rarely need
this, but understanding it demystifies performance and the famous "GIL". And the
tools to peek inside ship with Python.

## The pipeline: source → tokens → tree → bytecode → run

1. **Tokenizer** cuts source into tokens.
2. **Parser** builds an Abstract Syntax Tree (AST).
3. **Compiler** emits bytecode.
4. **Evaluator** runs the bytecode on a stack machine.

You can see the AST:

```python
import ast

tree = ast.parse("x = 2 + 3")
print(ast.dump(tree, indent=2))
```

## See the actual bytecode with `dis`

The `dis` module disassembles a function into the instructions Python runs.
Notice it literally does LOAD, multiply (BINARY_OP), RETURN.

```python
import dis

def add_and_double(a, b):
    return (a + b) * 2

dis.dis(add_and_double)
```

## Memory: reference counting

Every object tracks how many names point at it. When that count hits zero, the
memory is freed. (Cycles are cleaned by a separate garbage collector.)

```python
import sys

data = [1, 2, 3]
another_name = data
print("references to the list:", sys.getrefcount(data))   # a bit inflated by the call itself
```

## The GIL, in one breath

CPython runs only **one thread of Python bytecode at a time** (the Global
Interpreter Lock). So:
- **I/O-bound** work (waiting on files/network) still benefits from threads.
- **CPU-bound** work needs `multiprocessing`, vectorized NumPy, or a
  free-threaded build (Python 3.13+).

For this lab the rule is simple: **vectorize with NumPy before you reach for
threads.**
