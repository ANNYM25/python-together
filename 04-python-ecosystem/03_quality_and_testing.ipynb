## How we test in this lab

Every lesson that *implements* something ends with a `✔️ Quick check` cell of
`assert` statements. An `assert` is a line that must be true — if it isn't,
Python raises an error and the cell fails. Since our CI runs every notebook top
to bottom, a broken implementation fails the build automatically. No separate
test suite to remember.

````python
import math

def simple_return(start, end):
    return end / start - 1

# a quick check, exactly like the ones in the algorithm/ML lessons:
assert math.isclose(simple_return(100, 110), 0.10)
assert math.isclose(simple_return(50, 40), -0.20)
print("checks passed ✅")
````

## Gotcha: never test floats with `==`

Notice we used `math.isclose`, not `==`. Because computers store decimals in
binary, `110/100 - 1` is actually `0.10000000000000009`, **not** exactly `0.1`.
Comparing floats with `==` is a classic bug — compare with a tolerance instead.

````python
print("110/100 - 1 =", 110 / 100 - 1)
print("== 0.1 ?        ", (110 / 100 - 1) == 0.1)          # False! surprise
print("isclose 0.1 ?   ", math.isclose(110 / 100 - 1, 0.1))  # True, the right way
````

## Style we keep

- **4 spaces** per indent; `snake_case` for functions/variables, `PascalCase`
  for classes.
- **Readable names** over clever ones — these are lessons.
- Keep `try` blocks **small** so you know exactly which line failed.
- A linter (e.g. `ruff`) is optional and listed in `requirements-dev.txt`.
- Every lesson ends with a `✔️ Quick check` (if it implements something) and a
  `🏋️ Your turn`.
