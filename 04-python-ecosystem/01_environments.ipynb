## Where does Python look for imports?

When you `import something`, Python searches the paths in `sys.path`, in order:
the current folder first, then the environment's installed packages. Let's look.

````python
import sys

print("Python executable:", sys.executable)
print("\nfirst few import search paths:")
for p in sys.path[:5]:
    print("  ", p or "(current directory)")
````

That `sys.executable` is the key tell: if it points inside a `.venv/` folder,
you're in your project's isolated environment. If it points at a system Python,
you probably forgot to activate the venv.

## What's installed here?

````python
import importlib.metadata as md

for pkg in ["numpy", "pandas", "scikit-learn"]:
    try:
        print(f"{pkg:14} {md.version(pkg)}")
    except md.PackageNotFoundError:
        print(f"{pkg:14} not installed")
````

## The ritual (terminal, not a notebook cell)

```bash
python3 -m venv .venv            # create, once per project
source .venv/bin/activate        # Windows: .venv\Scripts\activate
pip install -r requirements.txt  # install into THIS env
```

Never commit the `.venv/` folder. `requirements.txt` records *what* to install;
the folder is disposable. In **Colab**, the environment is pre-built — you skip
all of this.
