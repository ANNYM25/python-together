# 1 · NumPy — fast maths on whole arrays

A 10,000-day return series should never be a Python `for` loop. **NumPy** gives
you the *array*: do maths on millions of numbers at once, in fast C under the
hood. Every data and ML library is built on it.

```python
import numpy as np
```

## Make an array and operate on all of it at once

Here: a year (252 trading days) of simulated daily returns. Notice we compute
mean and volatility across the whole array with no loop — this is
"vectorization".

```python
rng = np.random.default_rng(0)
rets = rng.normal(0.0004, 0.01, size=252)

print("shape:", rets.shape, " dtype:", rets.dtype)
print("mean daily return:", rets.mean().round(5))
print("daily volatility :", rets.std().round(5))
print("worst day        :", rets.min().round(4))
```

## Whole-array arithmetic ("broadcasting")

Multiply/add a scalar and it applies to every element. `np.cumprod` turns daily
returns into a growth curve.

```python
wealth = np.cumprod(1 + rets)
print("growth of £1 over the year:", wealth[-1].round(4))

# annualize: scale daily stats to a year
print("annualized mean:", (rets.mean() * 252).round(4))
print("annualized vol :", (rets.std() * np.sqrt(252)).round(4))
```

## Indexing and masks

A boolean "mask" selects elements by a condition — the foundation of filtering
data.

```python
down_days = rets[rets < 0]
print("number of down days:", down_days.size)
print("average down-day move:", down_days.mean().round(4))
```

## 2-D arrays (matrices)

```python
matrix = rng.normal(size=(5, 3))
print("column means:", matrix.mean(axis=0).round(3))   # axis=0 = down the columns
print("row means   :", matrix.mean(axis=1).round(3))   # axis=1 = across each row
```

### ✅ What you learned
- NumPy **arrays** do maths on whole collections at once (vectorization) — no loops.
- Scalar ops **broadcast**; `cumprod`, `mean`, `std`, `min` summarize.
- Boolean **masks** (`rets[rets < 0]`) filter data.
- `axis=0` works down columns, `axis=1` across rows.
