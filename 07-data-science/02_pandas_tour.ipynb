# 2 · pandas — labelled tables

Finance and ML both start as a **table**: rows indexed by date, columns of
features. **pandas** is NumPy with labels — a spreadsheet you drive with code.
The `DataFrame` is the single most-used object in all of data science.

```python
import numpy as np
import pandas as pd
```

## Build a DataFrame

A date index plus two columns of simulated asset returns.

```python
rng = np.random.default_rng(1)
idx = pd.date_range("2024-01-02", periods=8, freq="B")   # 8 business days
df = pd.DataFrame(
    {
        "equity": rng.normal(0.0005, 0.012, size=len(idx)),
        "bond":   rng.normal(0.0001, 0.003, size=len(idx)),
    },
    index=idx,
)
df.round(4)
```

## Create new columns from old ones

A 60/40 blend — just arithmetic on columns, applied row-wise automatically.

```python
df["blend"] = 0.6 * df["equity"] + 0.4 * df["bond"]
df.round(4)
```

## Summarize and relate

`.describe()` gives summary stats; `.corr()` shows how columns move together
(−1 to +1). Correlation is central to both portfolio risk and feature selection.

```python
print(df["equity"].describe().round(4))
print("\ncorrelation matrix:\n", df.corr().round(3))
```

## Rolling windows and selection

```python
print("3-day rolling equity vol:\n", df["equity"].rolling(3).std().round(4))
print("\ndays equity returned > 0:\n", df[df["equity"] > 0].index.date)
```
