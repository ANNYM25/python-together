# 3 · Visualization — a chart catches bugs a metric hides

A single number can lie; a picture rarely does. **matplotlib** is Python's
plotting workhorse. In a notebook, plots appear right under the cell — no files
to save. We'll plot an equity curve and a histogram of returns.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## A line plot: the growth of £1

Create a figure, plot on its axes, label everything, and `show()` it. Clear
labels and a title are not optional — a chart nobody can read is useless.

```python
rng = np.random.default_rng(2)
rets = rng.normal(0.0004, 0.01, size=252)
curve = np.cumprod(1 + rets)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(curve, color="#1f4e5f")
ax.set_title("Synthetic equity curve")
ax.set_xlabel("trading day")
ax.set_ylabel("growth of £1")
fig.tight_layout()
plt.show()
```

## A histogram: the shape of the returns

Line plots show *sequence*; histograms show *distribution* — how often each size
of move happens. This is how you eyeball whether data looks Gaussian (maths
track, lesson 4).

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(rets, bins=30, color="#8a5a44", edgecolor="white")
ax.set_title("Distribution of daily returns")
ax.set_xlabel("daily return")
ax.set_ylabel("number of days")
fig.tight_layout()
plt.show()
```

### ✅ What you learned
- `fig, ax = plt.subplots()` then `ax.plot(...)` / `ax.hist(...)` draws a chart.
- Always label axes and add a title.
- **Line** plots show sequence over time; **histograms** show distribution.
- In notebooks, `plt.show()` renders the figure inline.
