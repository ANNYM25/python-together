# 4 · Statistics — beyond the average

The mean is where most people stop. But the *spread*, the *tails*, and *how much
to trust a number from a limited sample* matter just as much — in ML (is my
model really better?) and finance (how bad is a bad day?). This lesson builds
the stats you'll reach for constantly.

```python
import numpy as np
```

## Center: mean vs median

The **mean** is the average; the **median** is the middle value. When they
disagree, the data is **skewed** (lopsided) — the mean gets dragged by extremes,
the median shrugs them off. We draw from a heavy-tailed (Student-t) distribution
to see it.

```python
rng = np.random.default_rng(3)
x = rng.standard_t(df=5, size=2000)

print("mean  :", x.mean().round(4))
print("median:", np.median(x).round(4))
print("std   :", x.std().round(4))
```

## Spread and tails: quantiles

A **quantile** is "the value below which this fraction of the data falls". The
5th percentile is a simple Value-at-Risk: "we expect to do worse than this only
5% of the time."

```python
q = np.quantile(x, [0.05, 0.5, 0.95])
print("5th / 50th / 95th percentile:", q.round(3))
```

## How much to trust the mean: standard error

A mean from a *sample* is itself uncertain. The **standard error** shrinks as
your sample grows (`std / √n`). Mean ± 2·SE is a rough 95% confidence interval —
the backbone of "is this result real or noise?"

```python
se = x.std(ddof=1) / np.sqrt(len(x))
print("standard error of the mean:", round(se, 4))
print(f"95% CI for the mean: [{x.mean() - 2*se:.3f}, {x.mean() + 2*se:.3f}]")
```

## When you don't know the formula: bootstrap

Resample the data *with replacement* many times, recompute the statistic each
time, and look at the spread. This estimates uncertainty for *any* statistic
without a formula — a genuinely powerful idea.

```python
boot_medians = [np.median(rng.choice(x, size=len(x), replace=True)) for _ in range(500)]
lo, hi = np.percentile(boot_medians, [5, 95])
print(f"bootstrap 90% interval for the median: [{lo:.3f}, {hi:.3f}]")
```

### ✅ What you learned
- **Mean vs median** reveals skew; tails matter, not just the center.
- **Quantiles** describe spread and tail risk (a simple VaR).
- **Standard error** (`std/√n`) says how much to trust a sample mean.
- The **bootstrap** estimates uncertainty for any statistic by resampling.
