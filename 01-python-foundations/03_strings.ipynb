# 3 · Strings — working with text

Tickers, report names, file paths, column headers, model labels — all text.
A "string" is just text wrapped in quotes. Let's bend it to our will.

## f-strings: dropping values into text

Put an `f` before the quotes, then anything in `{curly braces}` is replaced by
its value. This is the one string trick you'll use most.

```python
name = "Annie"
field = "sustainable finance"

line = f"{name} studies {field}."
print(line)
```

## Changing how text looks

Strings come with built-in helpers you call with a dot.

```python
print("UPPER:", line.upper())
print("lower:", line.lower())
print("title:", line.title())
```

## Slicing — grabbing part of a string

`text[start:stop]` takes characters from `start` up to (but not including)
`stop`. Counting starts at **0**.

```python
print("first 5 chars:", line[:5])    # characters 0,1,2,3,4
print("char at 0:", line[0])
```

## Searching and splitting

```python
print("contains 'finance'? ", "finance" in line)
print("split into words:", line.replace(".", "").split())
```

## Cleaning messy text

Real data is dirty. `.strip()` removes the blank space around text — a daily
chore when reading spreadsheets.

```python
ticker = "  esg.xlb  "
clean = ticker.strip().upper()
print(f"before: '{ticker}'")
print(f"after:  '{clean}'")
```

- `f"{value}"` drops values into text.
