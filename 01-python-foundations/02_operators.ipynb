# 2 · Operators — doing things with values

Models and money both start as arithmetic on numbers. Let's see every
operator you'll actually use. Run each cell as you read.

## Maths operators

We'll use two numbers, `a` and `b`.

```python
a = 7
b = 3

print("add       ", a + b)
print("subtract  ", a - b)
print("multiply  ", a * b)
print("divide    ", a / b)    # always gives a float
print("floor div ", a // b)   # divide and throw away the remainder
print("remainder ", a % b)    # what's left over ("modulo")
print("power     ", a ** b)   # a to the power of b
```

`//` and `%` go together: `a // b` is how many whole times `b` fits into `a`,
and `a % b` is what's left. Splitting money into equal chunks uses both.

## Comparisons — these give `True` or `False`

You'll lean on these constantly once we hit `if` statements and filtering data.

```python
print("equal?       ", a == b)   # note: two equals signs to COMPARE
print("not equal?   ", a != b)
print("greater?     ", a > b)
print("less or eq?  ", a <= b)
```

## Logic — combining True/False values

`and` needs both sides true. `or` needs at least one. `not` flips it.

```python
print("and", True and False)
print("or ", True or False)
print("not", not False)
```

## Shorthand that shows up everywhere

`wealth *= 1.05` means `wealth = wealth * 1.05`. These "compound assignment"
operators are how loops build up a running total.

```python
wealth = 1000.0
wealth *= 1.05          # grow by 5%
print("wealth after 5% growth:", wealth)
```
