# 4 · Lists — many values in a row

A return series, a mini-batch of training examples, and a shopping list are
all the *same shape*: an ordered collection of items. In Python that's a
**list** — square brackets, items separated by commas.

```python
returns = [0.01, -0.02, 0.015, 0.0, 0.004]
print(returns)
```

## Looking inside

Items are numbered from **0**. Negative numbers count from the end.

```python
print("how many:", len(returns))
print("first:", returns[0])
print("last:", returns[-1])
print("the middle chunk:", returns[1:4])   # items 1,2,3
```

## Lists can change

Unlike a lot of things in Python, a list can be edited in place.

```python
returns.append(0.008)    # add to the end
returns[1] = -0.018      # overwrite item 1
print("updated:", returns)
```

## Useful things Python does for you

```python
print("sum:", sum(returns))
print("biggest:", max(returns))
print("sorted copy:", sorted(returns))
print("original untouched:", returns)
```

## A sneak peek: list comprehensions

This compact line builds a *new* list by doing something to every item.
Read it as "two times r, for each r in returns". You'll use this constantly.

```python
doubled = [2 * r for r in returns]
print(doubled)
```

### ✅ What you learned
- `[a, b, c]` makes a list; `len(...)` counts it.
- Index from 0 (`list[0]`), from the end with negatives (`list[-1]`), slice `list[a:b]`.
- `.append(...)` adds; `list[i] = ...` overwrites.
- `sum / max / min / sorted` just work on lists.
- `[f(x) for x in things if condition]` builds a new list.
