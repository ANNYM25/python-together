## Values have a *type*

Every value in Python is a certain *kind* of thing. The ones you will meet
first are:

| Type    | What it is            | Example        |
|---------|-----------------------|----------------|
| `int`   | a whole number        | `10`           |
| `float` | a number with a dot   | `9.81`         |
| `str`   | text ("string")       | `"Anny"`      |
| `bool`  | True or False         | `True`         |

You can ask Python the type of any value with `type(...)`.

```python
print(type(10))
print(type(9.81))
print(type("Anny"))
print(type(True))
```

## Giving a value a name (a *variable*)

We rarely leave values floating. We give them names so we can reuse them.
The `=` sign means "store the thing on the right under the name on the left".

```python
name = "Anny"
started_coding = 2026
pi = 3.14159

print(name, "started in", started_coding)
print("pi is a", type(pi).__name__)
```

`type(pi).__name__` just gives the short name of the type (`float`) instead of
the longer `<class 'float'>`. Small nicety — don't worry about the `.__name__`
part yet.

## A first taste of real data

Because this lab leans toward **machine learning and finance**, here is a
value that actually means something: a tiny portfolio, written as a list of
daily returns (how much an investment moved each day, as a fraction).

```python
daily_returns = [0.02, -0.01, 0.015]
print(daily_returns)
print("that is a", type(daily_returns).__name__)
print("with", len(daily_returns), "days in it")
```

You will meet lists properly in lesson 4. For now just notice: a list is
*itself* a type, and it holds other values.
