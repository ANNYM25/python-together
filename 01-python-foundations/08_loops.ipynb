# 8 · Loops — doing something to every item

Almost every dataset is something you walk through once. A `for` loop does
exactly that: run the same block of code for each item in a collection.

```python
prices = [100, 101, 99, 102]

for price in prices:
    print("price:", price)
```

## When you also want the position number

`enumerate(...)` hands you the index *and* the item.

```python
for day, price in enumerate(prices):
    print("day", day, "->", price)
```

## Walking two things side by side

`zip(...)` pairs up items from two sequences. Here we pair each price with the
next one to compute **daily returns** (today ÷ yesterday − 1).

```python
simple_returns = []
for yesterday, today in zip(prices, prices[1:]):
    simple_returns.append(today / yesterday - 1)

print("returns:", [round(r, 4) for r in simple_returns])
```

## `range`, `while`, `break`, `continue`

- `range(n)` gives `0, 1, ... n-1` — a loop that runs a fixed number of times.
- `while` repeats as long as a condition stays true.
- `continue` skips to the next item; `break` stops the loop entirely.

```python
for x in range(5):
    if x == 2:
        continue      # skip 2
    if x == 4:
        break         # stop before 4
    print("x =", x)
```

```python
n = 3
while n > 0:
    print("countdown", n)
    n -= 1
```
