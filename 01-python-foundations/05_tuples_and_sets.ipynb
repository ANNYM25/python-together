# 5 · Tuples and sets — collections with rules

A list is flexible. Sometimes you want the opposite:
- a **tuple** = a sequence that *can't change* (round brackets),
- a **set** = a bag of *unique* items, no duplicates, no order (curly braces).

## Tuples: fixed records

The classic machine-learning split never changes mid-experiment, so a tuple
fits perfectly.

```python
split = ("train", "validate", "test")
print(split)
print("where is 'validate'?", split.index("validate"))
```

Because tuples can't change, Python lets you use them as dictionary keys
(lists can't do this). Handy for labelling data shapes.

```python
shapes = {(28, 28): "mnist image", (1, 10): "one-hot digit"}
print(shapes[(28, 28)])
```

## Sets: automatic de-duplication

Turn a messy list into a set and the duplicates vanish.

```python
holdings = ["bond", "equity", "equity", "cash", "bond"]
unique = set(holdings)
print("unique asset classes:", unique)
print("any alternatives?", "alternatives" in unique)
```

## Set maths

Sets answer "what's shared / combined / different" questions directly.

```python
a = {"equity", "bond"}
b = {"bond", "cash"}

print("in either (union):     ", a | b)
print("in both (intersection):", a & b)
print("only in a (difference):", a - b)
```

### ✅ What you learned
- Tuples `( )` are ordered and *unchangeable* — good for fixed records and dict keys.
- Sets `{ }` hold *unique* items with no order.
- Set operators: `|` union, `&` intersection, `-` difference, `^` exactly-one.
