# 6 · Dictionaries — labelled data

A row in a spreadsheet, a JSON response from an API, a model's settings —
they're all the same idea: **keys** that point to **values**. That's a
dictionary. Curly braces, `key: value` pairs.

```python
position = {"ticker": "MSFT", "shares": 12, "price": 410.5}
print(position["ticker"])    # look up by key
```

## Changing a dictionary

```python
position["shares"] += 3          # update an existing value
position["sector"] = "technology"  # add a brand-new key
print(position)
```

## Asking what's inside

```python
print("keys:", list(position.keys()))
print("values:", list(position.values()))
print("market value:", position["shares"] * position["price"])
```

## Dictionaries inside dictionaries

This is how real portfolios, JSON, and config files are shaped. We loop over
each entry with `.items()`, which hands us the key *and* the value together.

```python
portfolio = {
    "MSFT": {"shares": 12, "price": 410.5},
    "BND": {"shares": 40, "price": 72.1},
}

total = 0.0
for ticker, lot in portfolio.items():
    value = lot["shares"] * lot["price"]
    total += value
    print(ticker, "is worth", round(value, 2))

print("total portfolio value:", round(total, 2))
```
