# 7 · Conditionals — making decisions

Code that always does the same thing isn't very useful. `if` lets your program
choose a path based on the data. Risk rules, trading signals, and the branches
inside a model are all `if` statements underneath.

## The shape of an `if`

Python uses **indentation** (spaces at the start of a line) to show what's
"inside" the `if`. Four spaces is the convention.

```python
drawdown = -0.12

if drawdown < -0.10:
    print("review the book: drawdown past 10%")
elif drawdown < -0.05:          # "else if"
    print("watchlist")
else:
    print("inside tolerance")
```

Read it top to bottom: Python checks each condition in order and runs the
**first** one that's true. `elif` and `else` are optional.

## Bucketing a value

A very common pattern: turn a number into a category. Try changing `vol`
and re-running.

```python
vol = 0.12

if vol < 0.08:
    bucket = "low"
elif vol < 0.18:
    bucket = "medium"
else:
    bucket = "high"

print(f"volatility {vol} -> {bucket} risk")
```
