# 5 · Optimization — walking downhill

Fitting a model is almost always: define how wrong you are (a **loss**), then
adjust the parameters to make the loss smaller. The workhorse is **gradient
descent** — take a step in the downhill direction (the negative slope), repeat.
This lesson is the pure-maths version of what the ML track does to real models.

## A simple bowl-shaped loss

`loss(w) = (w − 3)²` is smallest at `w = 3`. Its slope (derivative) is
`2(w − 3)`. Gradient descent should find `w = 3` without being told.

```python
def loss(w):
    return (w - 3) ** 2

def grad(w):
    return 2 * (w - 3)       # the derivative: points uphill, so we step against it
```

## The descent loop

`w = w − learning_rate × slope`. The slope tells us which way is uphill; the
minus sign sends us downhill; the learning rate sets the step size.

```python
w = 0.0
lr = 0.2
for step in range(21):
    w -= lr * grad(w)
    if step % 5 == 0:
        print(f"step {step:02d}  w={w:.4f}  loss={loss(w):.4f}")

print("\nconverged to w =", round(w, 4), "(true minimum is 3)")
```

## The learning rate is a knob you must tune

Too small → painfully slow. Too big → it overshoots and **diverges**. Watch what
a too-large rate does:

```python
w = 0.0
lr_big = 1.1          # too large on purpose
for step in range(6):
    w -= lr_big * grad(w)
    print(f"step {step}  w={w:.3f}  loss={loss(w):.3f}")
print("^ bouncing away instead of settling — the step size is too big.")
```

This is the *same* algorithm that fits linear regression (ML lesson 2) and
trains the neural network (ML lesson 7) — just with many parameters and a
gradient computed by backprop.
