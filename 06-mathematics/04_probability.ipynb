# 4 · Probability — reasoning about uncertainty

Data is noisy and the future is uncertain, so ML and finance both run on
probability. A loss function is secretly an assumption about noise;
Value-at-Risk is a tail probability. We'll build intuition with simulation
(just generate lots of outcomes and count) plus a first look at the Gaussian.

```python
import math
import numpy as np
```

## Expectation — the probability-weighted average

`E[X]` is each outcome times its probability, summed. It's the "center of mass"
of a random quantity — the expected return of a bet.

```python
outcomes = np.array([-0.10, 0.00, 0.10])
probs    = np.array([0.20, 0.50, 0.30])
print("probabilities sum to:", probs.sum())
print("E[return] =", float(outcomes @ probs))     # dot product = weighted sum
```

## Simulation beats memorised formulas

Don't remember the formula? Simulate. Draw 10,000 samples from a normal
("Gaussian") distribution and just *measure* what you want.

```python
rng = np.random.default_rng(7)
draws = rng.normal(loc=0.01, scale=0.02, size=10_000)   # mean 1%, std 2%
print("sample mean:", round(float(draws.mean()), 4))
print("sample std :", round(float(draws.std()), 4))
print("P(loss, i.e. x<0):", round(float((draws < 0).mean()), 3))
```

## The Gaussian (normal) distribution

The famous bell curve. Its formula gives the *density* at a point; the area
under it between two points is a probability. Many natural quantities and
measurement errors look Gaussian.

```python
def gaussian_pdf(x, mu=0.0, sigma=1.0):
    z = (x - mu) / sigma
    return math.exp(-0.5 * z * z) / (sigma * math.sqrt(2 * math.pi))

for x in [-2, -1, 0, 1, 2]:
    bar = "#" * int(gaussian_pdf(x) * 100)
    print(f"x={x:+d}  {bar}")
```

## A peek at Bayes

Probability also lets you *update* a belief with evidence. `P(A|B) =
P(B|A)·P(A) / P(B)`. This underlies spam filters, medical tests, and Bayesian
ML — worth knowing the name now, we'll use the idea later.
