# 2 · Calculus with Python

A **derivative** is just a slope: how fast something changes. An **integral** is
just an area. You don't need to remember rules from school — Python can compute
both numerically, and `sympy` can do them symbolically. This is the exact maths
behind **gradient descent** (ML track, lesson 2 & 7): the gradient *is* a
derivative.

```python
import math
```

## A derivative is a slope — measure it numerically

The "finite difference" trick: the slope at `x` is approximately the rise over
run for a tiny step `h`. We check it against the known answer for
`f(x) = x³ − 2x`, whose derivative is `3x² − 2`.

```python
def finite_difference(fn, x, h=1e-6):
    return (fn(x + h) - fn(x - h)) / (2 * h)

def f(x):
    return x ** 3 - 2 * x

x0 = 1.5
print("numeric slope at 1.5:", round(finite_difference(f, x0), 5))
print("exact  slope (3x²−2):", round(3 * x0 ** 2 - 2, 5))
```

## An integral is an area — add up thin slices

The trapezoid rule chops the area under a curve into thin strips and sums them.
We check `∫sin(x) dx` from 0 to π, which is exactly 2.

```python
def trap_integral(fn, a, b, n=200):
    width = (b - a) / n
    total = 0.5 * (fn(a) + fn(b))
    for i in range(1, n):
        total += fn(a + i * width)
    return total * width

print("area under sin, 0..π:", round(trap_integral(math.sin, 0.0, math.pi), 5), "(true = 2)")
```

## Symbolic calculus with sympy

`sympy` does the algebra the way a textbook would — exact, not approximate.

```python
import sympy as sp

x = sp.symbols("x")
g = sp.exp(-x ** 2)
print("d/dx of exp(−x²) =", sp.diff(g, x))
print("∫ exp(−x²) dx from 0 to 1 ≈", sp.integrate(g, (x, 0, 1)).evalf())
```

## Why this matters for ML

Training a model = minimise a loss. To know which way is "downhill", you need
the **slope of the loss** with respect to each parameter — a derivative. That's
literally the `w -= lr * gradient` line in the ML track. Calculus isn't optional
background; it's the engine.
