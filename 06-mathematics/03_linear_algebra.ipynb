# 3 · Linear algebra — the maths of data

A dataset **is** a matrix: rows are examples, columns are features. A linear
model **is** a matrix-vector product. Get comfortable with vectors and matrices
and huge chunks of ML stop being mysterious. NumPy makes it concrete.

```python
import numpy as np
```

## Vectors and the dot product

A dot product multiplies matching entries and sums them — it's how a linear
model combines features with weights (`w·x`).

```python
weights = np.array([0.5, -0.2, 0.1])
features = np.array([2.0, 3.0, 10.0])
print("w · x =", weights @ features)     # @ is matrix/vector multiply
```

## Matrices and matrix-vector products

`A @ v` applies the matrix `A` to the vector `v`. `.T` transposes (flips rows
and columns). These two operations are 90% of what a neural network layer does.

```python
A = np.array([[1.0, 2.0],
              [3.0, 4.0]])
v = np.array([1.0, 1.0])
print("A @ v =", A @ v)
print("transpose A.T =\n", A.T)
print("determinant:", round(float(np.linalg.det(A)), 4))
```

## Least squares = fitting a line with linear algebra

Fitting `y ≈ a·x + b` to data is a linear-algebra problem NumPy solves in one
call. This is linear regression (ML lesson 2) done the closed-form way.

```python
x = np.array([1.0, 2.0, 3.0])
y = np.array([1.1, 1.9, 3.2])
X = np.column_stack([x, np.ones_like(x)])    # add a column of 1s for the intercept
coef, *_ = np.linalg.lstsq(X, y, rcond=None)
print("slope, intercept =", coef.round(4))
```
