# 1 · Algebra you actually use

The maths track uses Python as a calculator with superpowers — so the symbols
become something you can *run*. We start with the algebra hiding under every
finance and ML formula: powers, logs, and geometric series (the engine of
compound interest and discounting).

```python
import math
```

## Powers and logs are inverses

`log` undoes `exp`. This matters because **log returns** add up nicely over time
where simple returns don't.

```python
print("log of 1.1:", round(math.log(1.1), 4))
print("exp undoes log:", round(math.exp(math.log(1.1)), 4))
print("a 10% gain as a log return:", round(math.log(110 / 100), 4))
```

## Geometric series — the shape of money over time

`a + a·r + a·r² + ... + a·r^(n-1)` sums to `a·(1 − rⁿ)/(1 − r)`. Discounting a
stream of future cash flows is exactly this sum.

```python
def geometric_sum(a, r, n):
    if n <= 0:
        return 0.0
    if r == 1:
        return a * n
    return a * (1 - r ** n) / (1 - r)

# Present value of 100/year for 10 years, discounted at 5%
payment, rate, years = 100.0, 0.05, 10
pv = geometric_sum(payment / (1 + rate), 1 / (1 + rate), years)
print("present value of the annuity:", round(pv, 2))
```

## Solving with logs

"How long to double my money at 4%?" means solving `1.04ⁿ = 2`. Take logs of
both sides: `n = log(2) / log(1.04)`.

```python
n = math.log(2) / math.log(1.04)
print("years to double at 4%:", round(n, 2))
```
