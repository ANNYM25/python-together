# 4 · ESG and sustainable finance

Sustainable finance asks: not just *what return*, but *at what cost to the
planet and society*. **ESG** scores (Environmental, Social, Governance) try to
quantify that — and an ESG score is really *a model with a policy wrapped around
it*. We'll build a tiny, transparent one. (All issuers and numbers are made up.)

```python
from dataclasses import dataclass
```

## Model an issuer

Each issuer gets E, S, G sub-scores (0–100, higher = better here) plus a hard
flag — e.g. fossil-fuel exposure — that can override the score entirely.

```python
@dataclass
class Issuer:
    name: str
    e: float
    s: float
    g: float
    fossil_flag: bool

book = [
    Issuer("North Wind",   82, 70, 65, False),
    Issuer("City Transit", 74, 80, 60, False),
    Issuer("Old Refinery", 40, 55, 50, True),
    Issuer("Agri Co-op",   60, 78, 58, False),
]
```

## A blended score (a weighted policy choice)

The weights are a *value judgement*, not a fact — weighting E more heavily is a
choice. That's the "policy wrapped around a model" point.

```python
def blended_score(issuer, weights=(0.4, 0.3, 0.3)):
    we, ws, wg = weights
    return we * issuer.e + ws * issuer.s + wg * issuer.g

for issuer in book:
    print(f"{issuer.name:14} blended ESG: {blended_score(issuer):.1f}")
```

## An eligibility rule (screening)

A fund might require: no fossil flag **and** a blended score above a floor. This
is exactly how a sustainable fund decides what it may hold.

```python
def eligible(issuer, floor=55.0):
    if issuer.fossil_flag:
        return False
    return blended_score(issuer) >= floor

for issuer in book:
    print(f"{issuer.name:14} eligible? {eligible(issuer)}")
```

## A toy green bond

Proceeds earmarked for, say, a wind project. Its cash flows are an ordinary
bond's — coupons then principal — which you can value with the tools from
lesson 1.

```python
face, coupon, years = 100.0, 0.032, 5
cash = [coupon * face] * (years - 1) + [coupon * face + face]
print("green bond cash flows:", cash)
```
