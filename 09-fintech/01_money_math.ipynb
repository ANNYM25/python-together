# 1 · Time value of money

The foundation of all finance: **£1 today is worth more than £1 next year**,
because today's pound can be invested. Discounting future cash to today, and
judging whether a project is worth it, all flow from this one idea. (All numbers
here are illustrative — no investment advice.)

## Present value — compound interest, run backwards

To grow money you multiply by `(1 + rate)` each year. To find today's value of
future money, you *divide*. That's a discount.

```python
def present_value(cashflow, rate, years):
    return cashflow / (1 + rate) ** years

print("£100 in 5 years, discounted at 6%, is worth today:",
      round(present_value(100, 0.06, 5), 2))
```

## Net Present Value (NPV) — is a project worth it?

Sum the present value of every cash flow, including the upfront cost at time 0
(usually negative). **NPV > 0 means the project creates value** at that discount
rate.

```python
def npv(rate, cashflows):
    return sum(cf / (1 + rate) ** t for t, cf in enumerate(cashflows))

project = [-1000, 300, 350, 400, 280]     # spend 1000 now, get cash back over 4 years
print("NPV at 8%:", round(npv(0.08, project), 2))
```

## Internal Rate of Return (IRR) — the break-even rate

The IRR is the discount rate that makes NPV exactly zero — the project's
effective annual return. We find it by **bisection**: repeatedly halve the range
where NPV crosses zero. (This reuses the "halve the search space" idea from the
algorithms track.)

```python
def irr(cashflows, lo=-0.9, hi=1.0, steps=80):
    for _ in range(steps):
        mid = (lo + hi) / 2
        if npv(mid, cashflows) > 0:
            lo = mid          # still positive: the break-even rate is higher
        else:
            hi = mid
    return (lo + hi) / 2

print("IRR of the project:", round(irr(project), 4), "(~", round(irr(project)*100, 1), "% )")
```

- **Present value** discounts future cash to today (compounding in reverse).
