# 3 · Portfolios — weights are a vector, risk is a matrix

A portfolio is a set of **weights** summing to 1. Its return is easy (a weighted
average). Its *risk* is the subtle part: it depends on how assets move
*together* — the **covariance** — not just each one's volatility. This is where
the linear-algebra track pays off. (Synthetic; not advice.)

```python
import numpy as np
```

## Simulate two assets that move together a little

A `multivariate_normal` lets us set each asset's mean, variance, and their
covariance. Think "equity" and "bond".

```python
rng = np.random.default_rng(9)
rets = rng.multivariate_normal(
    mean=[0.0004, 0.00015],
    cov=[[0.00015, 0.00002],
         [0.00002, 0.00003]],
    size=800,
)
mean = rets.mean(axis=0)
cov = np.cov(rets, rowvar=False)
print("mean daily returns:", mean.round(5))
print("covariance matrix:\n", cov.round(6))
```

## Portfolio return and volatility

Return is `weights · mean`. Volatility uses the covariance matrix:
`sqrt(wᵀ · cov · w)` — the quadratic form that captures diversification. Both
are annualized by scaling with 252 trading days.

```python
def port_stats(weights, mean, cov, periods=252):
    mu = float(weights @ mean) * periods
    vol = float(np.sqrt(weights @ cov @ weights) * np.sqrt(periods))
    return mu, vol
```

## Sweep the equity/bond mix

Walk the weight from 100% bonds to 100% equity and watch risk and return trade
off — the heart of asset allocation.

```python
print(" w_equity | ann_return | ann_vol")
for w_eq in np.linspace(0, 1, 11):
    w = np.array([w_eq, 1 - w_eq])
    mu, vol = port_stats(w, mean, cov)
    print(f"   {w_eq:4.2f}   |   {mu:6.4f}   | {vol:6.4f}")
```
