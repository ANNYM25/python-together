# 2 · Returns and risk

A headline "+12% this year" can hide a terrifying ride. Risk is about the
*path*, not just the endpoint. We'll measure volatility, the worst peak-to-
trough fall (drawdown), and a simple Value-at-Risk. (Synthetic data; not advice.)

```python
import numpy as np
```

## Simple vs log returns

Simple returns are intuitive; **log returns** (`log(1+r)`) add up cleanly over
time and are nicer for maths. For small daily moves they're almost identical.

```python
rng = np.random.default_rng(8)
rets = rng.normal(0.0005, 0.012, size=252)     # a year of daily returns
log_rets = np.log1p(rets)

print("simple mean daily:", rets.mean().round(5))
print("log    mean daily:", log_rets.mean().round(5))
print("annualized volatility:", (rets.std() * np.sqrt(252)).round(4))
```

## Drawdown — the pain metric

Maximum drawdown is the worst drop from a previous peak. Two strategies can have
the same average return but wildly different drawdowns — and drawdown is what
makes investors panic and sell.

```python
prices = 100 * np.cumprod(1 + rets)
peaks = np.maximum.accumulate(prices)     # the running high-water mark
drawdown = prices / peaks - 1.0
print("maximum drawdown:", round(float(drawdown.min()), 4),
      f"({drawdown.min()*100:.1f}%)")
```

## Value-at-Risk (VaR) — a tail number

The 5% VaR is the daily loss you'd expect to exceed only 1 day in 20. It's just
the 5th percentile of returns (connects to the stats track's quantiles).

```python
var_5 = float(np.quantile(rets, 0.05))
print("daily 5% VaR (as a return):", round(var_5, 4))
```

### ✅ What you learned
- **Log returns** add over time; simple returns are intuitive.
- **Volatility** (annualized std) measures the size of the wiggle.
- **Max drawdown** is the worst peak-to-trough fall — the pain metric.
- **VaR** is a tail loss = a low quantile of returns.
