# 4 · Linked lists — when "next" is a pointer

A Python `list` is great, but it hides how data connects in memory. A **linked
list** makes it explicit: each **node** holds a value and a reference to the
*next* node. It's the simplest structure where you manage the links yourself —
and the gateway to trees and graphs.

## A node

A `@dataclass` gives us a tiny class with an auto-generated `__init__` and
`__repr__`. Each node points to the next (or to `None` at the end).

```python
from dataclasses import dataclass
from typing import Optional

@dataclass
class Node:
    value: int
    next: Optional["Node"] = None

n = Node(1, Node(2, Node(3)))     # 1 -> 2 -> 3
print(n)
```

## A list that wraps the nodes

`prepend` adds to the front (cheap — just make a new head). `to_list` walks the
chain following `.next` until it hits `None`.

```python
class LinkedList:
    def __init__(self):
        self.head = None

    def prepend(self, value):
        self.head = Node(value, self.head)   # new node points at old head

    def to_list(self):
        out = []
        node = self.head
        while node is not None:
            out.append(node.value)
            node = node.next                 # follow the link
        return out

chain = LinkedList()
for value in (3, 2, 1):
    chain.prepend(value)
print(chain.to_list())
```

## ✔️ Quick check

```python
ll = LinkedList()
for v in (3, 2, 1):
    ll.prepend(v)
assert ll.to_list() == [1, 2, 3]
assert LinkedList().to_list() == []
print("all checks passed ✅")
```
