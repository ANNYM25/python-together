# 6 · Graphs — networks and how to explore them

A **graph** is nodes connected by edges. Knowledge graphs, transaction networks,
social connections, and the computation graph inside a neural network are all
graphs. Two fundamental ways to explore one: **breadth-first** (level by level,
using a queue) and **depth-first** (as deep as possible first, using recursion/
a stack). This is where the last two lessons pay off.

## Representing a graph

The simplest representation: a dict mapping each node to its neighbours (an
"adjacency list").

```python
graph = {
    "cash":      ["bonds", "equity"],
    "bonds":     ["ig_credit"],
    "equity":    ["climate", "tech"],
    "ig_credit": [],
    "climate":   [],
    "tech":      [],
}
```

## Breadth-first search (BFS) — a queue

Visit the start, then all its neighbours, then *their* neighbours — expanding in
rings. Uses a **queue** (FIFO) and a `seen` set so we never revisit.

```python
from collections import deque

def bfs(graph, start):
    seen = {start}
    order = [start]
    q = deque([start])
    while q:
        node = q.popleft()
        for neighbour in graph.get(node, []):
            if neighbour not in seen:
                seen.add(neighbour)
                order.append(neighbour)
                q.append(neighbour)
    return order

print("BFS from cash:", bfs(graph, "cash"))
```

## Depth-first search (DFS) — recursion

Go as deep as you can down one path before backtracking. Naturally **recursive**
(an implicit stack of function calls).

```python
def dfs(graph, start, seen=None, order=None):
    if seen is None:
        seen, order = set(), []
    seen.add(start)
    order.append(start)
    for neighbour in graph.get(start, []):
        if neighbour not in seen:
            dfs(graph, neighbour, seen, order)
    return order

print("DFS from cash:", dfs(graph, "cash"))
```

## ✔️ Quick check

```python
assert bfs(graph, "cash")[0] == "cash"
assert set(bfs(graph, "cash")) == set(graph.keys())    # reaches everything
assert dfs(graph, "equity") == ["equity", "climate", "tech"]
print("all checks passed ✅")
```

### ✅ What you learned
- A **graph** = nodes + edges; an adjacency-list dict is an easy representation.
- **BFS** explores in rings using a **queue**; great for shortest hops.
- **DFS** plunges deep using **recursion**; great for "can I reach / is there a path".
- A `seen` set prevents infinite loops on cycles.
