# 5 · Stacks and queues — order matters

Two of the most useful structures, defined by *how you remove things*:
- **Stack** — Last In, First Out (LIFO). Like a stack of plates. Undo, function
  calls, depth-first search.
- **Queue** — First In, First Out (FIFO). Like a line at a counter. Task
  scheduling, breadth-first search.

## A stack from a plain list

`append` pushes on top; `pop` takes from the top. The last thing in is the first
thing out — exactly how "undo" works.

```python
stack = []
stack.append("open file")
stack.append("type text")
stack.append("bold it")
print("undo:", stack.pop())      # removes "bold it"
print("undo:", stack.pop())      # removes "type text"
print("remaining:", stack)
```

## A queue from `collections.deque`

A `deque` ("deck") pops from the front cheaply (a plain list is slow at that).
`append` joins the back; `popleft` serves the front.

```python
from collections import deque

queue = deque()
queue.append("client A")
queue.append("client B")
queue.append("client C")
print("serve:", queue.popleft())   # A was first in, so first served
print("serve:", queue.popleft())   # then B
print("waiting:", list(queue))
```

## ✔️ Quick check

```python
s = []
for x in [1, 2, 3]:
    s.append(x)
assert s.pop() == 3 and s.pop() == 2     # LIFO

q = deque([1, 2, 3])
assert q.popleft() == 1 and q.popleft() == 2   # FIFO
print("all checks passed ✅")
```
