# 2 · Sorting — divide, conquer, combine

Sorting is the classic stage for the most important algorithmic idea:
**divide and conquer** — split a hard problem into smaller copies of itself,
solve those, and combine. We'll build merge sort and quicksort.

(In real code you just call `sorted(...)`. We build these to *understand* them.)

## Merge sort

Split the list in half, sort each half (by the same method — recursion!), then
**merge** the two sorted halves by repeatedly taking the smaller front item.
O(n log n).

```python
def merge(left, right):
    out = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            out.append(left[i]); i += 1
        else:
            out.append(right[j]); j += 1
    out.extend(left[i:])
    out.extend(right[j:])
    return out

def merge_sort(items):
    if len(items) <= 1:
        return items[:]
    mid = len(items) // 2
    left = merge_sort(items[:mid])      # sort the left half
    right = merge_sort(items[mid:])     # sort the right half
    return merge(left, right)           # combine

print(merge_sort([5, 1, 4, 2, 8, 5, 0]))
```

## Quicksort

Pick a **pivot**, split into "less / equal / greater", sort the two outer piles
recursively, and concatenate. Elegant, and fast on average.

```python
def quicksort(items):
    if len(items) <= 1:
        return items[:]
    pivot = items[len(items) // 2]
    less = [x for x in items if x < pivot]
    equal = [x for x in items if x == pivot]
    more = [x for x in items if x > pivot]
    return quicksort(less) + equal + quicksort(more)

print(quicksort([5, 1, 4, 2, 8, 5, 0]))
```

## ✔️ Quick check

```python
sample = [3, 1, 2, 3, 0]
assert merge_sort(sample) == [0, 1, 2, 3, 3]
assert quicksort(sample) == [0, 1, 2, 3, 3]
assert sample == [3, 1, 2, 3, 0]          # originals untouched (we copied)
print("all checks passed ✅")
```
