# 3 · Recursion and a taste of dynamic programming

**Recursion** is a function that calls itself on a smaller input until it hits a
simple "base case". You just used it in merge sort and quicksort. Trees,
factorials, and even neural-net backprop are recursive ideas.

## The anatomy: base case + recursive step

`factorial(n) = n × factorial(n-1)`, stopping at `factorial(1) = 1`. Without a
base case, it would recurse forever.

```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be >= 0")
    if n in (0, 1):          # base case: stop here
        return 1
    return n * factorial(n - 1)   # recursive step: smaller problem

print("5! =", factorial(5))
```

## When naive recursion is slow

Naive recursive Fibonacci recomputes the same values exponentially. The fix is
**dynamic programming**: build the answer bottom-up, reusing earlier results.
Here it's O(n) time and O(1) memory.

```python
def fib_dp(n):
    if n < 0:
        raise ValueError("n must be >= 0")
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b          # carry the last two forward
    return a

print("fib(10) =", fib_dp(10))
print("first 10 fibs:", [fib_dp(i) for i in range(10)])
```

> You saw the *other* fix in the pythonic-tools track: `@lru_cache` on the
> recursive version memoizes results automatically. DP and memoization are two
> routes to the same destination.

## ✔️ Quick check

```python
assert factorial(5) == 120
assert fib_dp(10) == 55
assert [fib_dp(i) for i in range(7)] == [0, 1, 1, 2, 3, 5, 8]
print("all checks passed ✅")
```
