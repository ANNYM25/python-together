# 1 · Searching — finding a needle

The algorithms track is about *how* computers do things efficiently, and *how to
reason about speed*. We start with search. Two ways to find a value in a list —
and a dramatic lesson in why being clever about structure pays off.

## Linear search: check every item

Simple, works on any list, but slow — on a list of `n` items it may look at all
`n`. We say it's **O(n)**.

```python
def linear_search(items, target):
    for index, value in enumerate(items):
        if value == target:
            return index
    return -1

xs = [3, 1, 4, 1, 5, 9]
print("find 5 at index:", linear_search(xs, 5))
print("find 2 (missing):", linear_search(xs, 2))
```

## Binary search: halve the problem each step

If the list is **sorted**, you can jump to the middle and throw away half the
list every comparison. On a million items, linear search may take a million
steps; binary search takes about 20. That's **O(log n)** — the payoff of
structure.

```python
def binary_search(sorted_items, target):
    lo, hi = 0, len(sorted_items) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        guess = sorted_items[mid]
        if guess == target:
            return mid
        if guess < target:
            lo = mid + 1      # target is in the right half
        else:
            hi = mid - 1      # target is in the left half
    return -1

ys = sorted(xs)
print("sorted:", ys)
print("binary find 5:", binary_search(ys, 5))
print("binary find 2 (missing):", binary_search(ys, 2))
```

## ✔️ Quick check

Running this cell verifies the implementations. (In this repo, running the
notebook *is* the test — these asserts must pass.)

```python
assert linear_search([4, 1, 9, 7], 9) == 2
assert linear_search([4, 1, 9, 7], 3) == -1
assert binary_search([1, 4, 7, 9], 7) == 2
assert binary_search([], 1) == -1
print("all checks passed ✅")
```
