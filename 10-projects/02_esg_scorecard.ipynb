# Project · An ESG scorecard

Build on the fintech ESG lesson: rank a book of (synthetic) issuers by a blended
ESG score and print a clean, sorted scorecard with an eligibility flag. A small
but real analyst deliverable. (All issuers invented; not advice.)

```python
from dataclasses import dataclass
```

## The scoring logic

Same transparent model as `09-fintech/04`: E/S/G sub-scores, a policy weighting,
a hard fossil-fuel screen, and a score floor.

```python
@dataclass
class Issuer:
    name: str
    e: float
    s: float
    g: float
    fossil_flag: bool

def blended_score(issuer, weights=(0.4, 0.3, 0.3)):
    we, ws, wg = weights
    return we * issuer.e + ws * issuer.s + wg * issuer.g

def eligible(issuer, floor=55.0):
    return (not issuer.fossil_flag) and blended_score(issuer) >= floor
```

## The book of issuers

```python
issuers = [
    Issuer("Harbor Solar",  88, 72, 70, False),
    Issuer("River Housing", 71, 90, 66, False),
    Issuer("Metals Inc",    48, 52, 61, False),
    Issuer("Legacy Coal",   22, 40, 44, True),
]
```

## The scorecard — ranked best to worst

`sorted(..., key=blended_score, reverse=True)` uses the higher-order-function
idea: pass the scoring function as the sort key.

```python
ranked = sorted(issuers, key=blended_score, reverse=True)

print(f"{'issuer':16}{'score':>8}{'eligible':>10}")
print("-" * 34)
for issuer in ranked:
    print(f"{issuer.name:16}{blended_score(issuer):8.1f}{str(eligible(issuer)):>10}")
```
