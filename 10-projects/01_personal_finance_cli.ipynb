# Project · A personal-finance ledger

Your first shipped thing. A tiny **ledger**: add labelled amounts (income
positive, spending negative), then print a running balance. It ties together
classes, methods, f-strings, and formatting — all from the earlier tracks.

```python
from dataclasses import dataclass, field
```

## The Ledger class

`rows` holds `(label, amount)` pairs. `field(default_factory=list)` gives each
new ledger its own empty list. Methods add entries, total the balance, and print
a tidy report.

```python
@dataclass
class Ledger:
    rows: list = field(default_factory=list)

    def add(self, label, amount):
        self.rows.append((label, amount))

    def balance(self):
        return sum(amount for _, amount in self.rows)

    def report(self):
        for label, amount in self.rows:
            sign = "+" if amount >= 0 else "-"
            print(f"{label:20} {sign}{abs(amount):8.2f}")
        print("-" * 30)
        print(f"{'balance':20}  {self.balance():8.2f}")
```

## Use it

```python
book = Ledger()
book.add("salary", 3200)
book.add("rent", -1100)
book.add("groceries", -240)
book.add("index fund", -400)
book.report()
```
