# 1 · What *is* machine learning?

A normal program is rules you write by hand: `if income > 50000 and ...`.
**Machine learning flips that around:** you show the computer *examples*, and it
finds the rule itself. The rule is a function `f(x) ≈ y` with numbers
("parameters") that get tuned until the predictions are good.

This lesson gives you the whole mental map before we build anything. A little
code runs under each idea so it's concrete, not just words.

```python
import numpy as np
```

## 1. Supervised vs unsupervised  (a.k.a. "labelled vs unlabelled")

The first fork in the road is: **do you know the right answer for each example?**

- **Supervised** — every row comes with its answer `y` (the *label*). You learn
  to predict `y` from the features `x`. This is what people usually mean by ML.
- **Unsupervised** — no answers. You look for structure: which rows are similar,
  what the main axes of variation are.

(People sometimes say "classified / unclassified data" — same idea: do we have
labels or not.)

```python
# Supervised: features x come WITH an answer y.
x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([0,   0,   1,   1])      # <- the labels we want to predict
print("supervised: we know the answer for each x:", list(zip(x, y)))

# Unsupervised: same x, but NO y. We'd look for groups ourselves.
print("unsupervised: just the x's, find structure:", x)
```

## 2. Inside supervised: regression vs classification

- **Regression** predicts a **number** — a price, a return, a temperature.
- **Classification** predicts a **category** — default / no-default, cat / dog.

```python
examples = {
    "predict tomorrow's stock price": "regression (a number)",
    "will this loan default?":        "classification (yes / no)",
    "how many bikes rented today?":   "regression (a count)",
    "is this email spam?":            "classification (spam / not)",
}
for task, kind in examples.items():
    print(f"- {task:35} -> {kind}")
```

## 3. Structured vs unstructured data

- **Structured** = neat tables of numbers/categories (a CSV, a SQL table).
  Classic ML (the next few lessons) is brilliant at this.
- **Unstructured** = raw text, images, audio. This is where **deep learning**
  wins — images use CNNs (lesson 08), text uses the transformers behind LLMs
  (your `10-projects/` tiny language model).

```python
structured = np.array([[40000, 2, 0.31],      # income, dependents, debt-ratio
                       [85000, 0, 0.12]])
print("structured (rows x features):\n", structured)
print("\nunstructured example (text):", repr("the market rallied on friday"))
```

## 4. The one rule that matters: generalization

We don't care how well a model does on data it already saw. We care how it does
on **new** data. So we always split our data into a **training** set (to learn
from) and a **test** set (held back, to judge honestly).

```python
rng = np.random.default_rng(0)
data = np.arange(10)
rng.shuffle(data)

train, test = data[:8], data[8:]      # 80% learn, 20% judge
print("train on:", train)
print("judge on:", test, "  <- the model never sees these while learning")
```

## 5. Overfitting — the classic trap

A model that is too flexible can *memorize* the training data, noise and all,
and then fail on new data. Below, a high-degree polynomial nails the training
points but goes wild between them. That gap — great on train, bad on test — is
**overfitting**.

```python
xs = np.linspace(0, 1, 8)
ys = np.sin(2 * np.pi * xs) + rng.normal(0, 0.1, size=8)

simple = np.polyfit(xs, ys, deg=1)     # a straight line (underfit)
complex_ = np.polyfit(xs, ys, deg=7)   # wiggles through every point (overfit)

train_err_simple = np.mean((np.polyval(simple, xs) - ys) ** 2)
train_err_complex = np.mean((np.polyval(complex_, xs) - ys) ** 2)
print(f"training error  — simple line: {train_err_simple:.3f}")
print(f"training error  — degree-7   : {train_err_complex:.3f}  (looks better!)")
print("...but the degree-7 curve swings wildly between points and fails on new x.")
```
