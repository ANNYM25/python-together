# 3 · Classification and metrics — why accuracy lies

When you predict a **category**, "what fraction did I get right?" (accuracy)
sounds like the obvious score. But it's dangerous. If only 2% of loans default,
a lazy model that screams "no default!" every time is **98% accurate** and
**completely useless**. This lesson gives you the metrics that don't lie.

```python
import numpy as np
```

## The confusion matrix — four kinds of outcome

For a yes/no (1/0) problem, every prediction lands in one of four boxes:

|                | predicted **yes** | predicted **no** |
|----------------|-------------------|------------------|
| **actually yes** | true positive (TP) | false negative (FN) |
| **actually no**  | false positive (FP) | true negative (TN) |

```python
y_true = np.array([0, 0, 0, 0, 0, 1, 1, 1])   # the truth
y_hat  = np.array([0, 0, 0, 1, 0, 1, 0, 1])   # the model's guesses

tp = int(np.sum((y_true == 1) & (y_hat == 1)))
tn = int(np.sum((y_true == 0) & (y_hat == 0)))
fp = int(np.sum((y_true == 0) & (y_hat == 1)))
fn = int(np.sum((y_true == 1) & (y_hat == 0)))
print(f"TP={tp}  FP={fp}  FN={fn}  TN={tn}")
```

## Precision and recall

- **Precision** = of everything we *flagged*, how much was right? `TP / (TP+FP)`
  → punishes false alarms.
- **Recall** = of everything that *was* positive, how much did we catch?
  `TP / (TP+FN)` → punishes misses.

There's a tension: catch more (recall ↑) and you usually raise false alarms
(precision ↓). Which matters more depends on the cost — missing a fraud vs.
annoying a good customer.

```python
precision = tp / (tp + fp)
recall    = tp / (tp + fn)
accuracy  = (tp + tn) / y_true.size
print(f"accuracy  {accuracy:.3f}")
print(f"precision {precision:.3f}")
print(f"recall    {recall:.3f}")
```

The same numbers from scikit-learn, to show the hand math matches the library:

```python
from sklearn.metrics import precision_score, recall_score
print("sklearn precision", round(precision_score(y_true, y_hat), 3))
print("sklearn recall   ", round(recall_score(y_true, y_hat), 3))
```

### ✅ What you learned
- Accuracy is misleading when classes are **imbalanced** (rare positives).
- The **confusion matrix** splits results into TP / FP / FN / TN.
- **Precision** = few false alarms; **recall** = few misses; **F1** balances them.
