# 6 · Unsupervised learning — structure without labels

Often you have features but **no answers** — customers you want to segment,
market days you want to group into "regimes", images you want to compress. Two
workhorses: **PCA** (squeeze many features into a few) and **k-means** (group
similar rows). No `y` anywhere in this lesson.

```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

## Data with hidden groups

We mix two "styles" of behaviour. Pretend we *don't* know which is which — the
point is to recover the groups from the numbers alone.

```python
rng = np.random.default_rng(5)
style_a = rng.normal(0.01, 0.02, size=(40, 3))
style_b = rng.normal(-0.005, 0.03, size=(40, 3))
X = np.vstack([style_a, style_b])
print("X shape:", X.shape, "(80 rows, 3 features, NO labels)")
```

## PCA — compress 3 features into 2

PCA finds the directions along which the data varies most, and lets you keep
just the top few. The `explained_variance_ratio_` tells you how much
information each kept axis carries.

```python
pca = PCA(n_components=2, random_state=0)
Z = pca.fit_transform(X)
print("kept 2 of 3 axes; variance explained:", pca.explained_variance_ratio_.round(3))
print("total info kept:", round(pca.explained_variance_ratio_.sum(), 3))
```

## k-means — group the rows

k-means picks `k` centres and assigns each point to its nearest centre, nudging
the centres until they settle. We ask for 2 clusters and see if it rediscovers
our two styles.

```python
labels = KMeans(n_clusters=2, n_init=10, random_state=0).fit_predict(Z)
counts = {int(c): int((labels == c).sum()) for c in np.unique(labels)}
print("cluster sizes:", counts, " (we built 40 + 40)")
```

## How many clusters? The "elbow"

k-means won't tell you the right `k`. One guide: plot the **inertia** (how tight
the clusters are) for several `k` and look for the "elbow" where extra clusters
stop helping much.

```python
for k in range(1, 6):
    km = KMeans(n_clusters=k, n_init=10, random_state=0).fit(Z)
    print(f"k={k}: inertia {km.inertia_:.2f}")
```

- Unsupervised learning finds structure with **no labels**.
