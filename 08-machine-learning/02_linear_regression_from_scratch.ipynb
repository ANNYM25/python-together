# 2 · Linear regression from scratch

Your first real model. The goal: fit a straight line `y = w·x + b` to data by
finding the best slope `w` and intercept `b`. We won't call a library — we'll
teach the line to fit itself with **gradient descent**, the same engine that
trains neural networks and LLMs.

```python
import numpy as np
```

## Make some data with a known answer

We invent data from a *true* line (slope 1.7, intercept −0.4) plus noise. If our
training works, it should rediscover roughly those numbers.

```python
rng = np.random.default_rng(4)
x = rng.normal(size=80)
y = 1.7 * x - 0.4 + rng.normal(scale=0.2, size=80)
print("first few x:", x[:3].round(2))
print("first few y:", y[:3].round(2))
```

## The idea of gradient descent

1. Start with a guess (`w = 0, b = 0`).
2. Measure how wrong we are — the **loss** (mean squared error).
3. Nudge `w` and `b` in the direction that lowers the loss.
4. Repeat.

The "direction that lowers the loss" is the **gradient** (a derivative — hello,
calculus track). For MSE the math works out to the two update lines below.

```python
w, b = 0.0, 0.0
lr = 0.1          # learning rate: how big a step we take
n = len(x)

for step in range(401):
    pred = w * x + b
    err = pred - y
    # gradients of mean-squared-error w.r.t. w and b:
    w -= lr * (2 / n) * np.dot(err, x)
    b -= lr * (2 / n) * err.sum()
    if step % 100 == 0:
        loss = np.mean(err ** 2)
        print(f"step {step:3d}  loss {loss:.4f}  w {w:.3f}  b {b:.3f}")

print(f"\nlearned: w={w:.3f}, b={b:.3f}   (true was 1.7, -0.4)")
```

Watch the loss fall and `w`, `b` crawl toward the true values. That's learning —
no magic, just repeated small downhill steps.
