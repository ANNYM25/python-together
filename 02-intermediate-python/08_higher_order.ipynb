# 8 · Higher-order functions — functions that take functions

In Python, functions are just values: you can store them, pass them, and return
them. A function that takes or returns another function is **higher-order**. This
idea is everywhere in data work — `df.apply(f)`, `sorted(key=f)`, sklearn
scorers all take a function as an argument.

```python
rets = [0.01, -0.02, 0.015]
```

## `map` — apply a function to every item

```python
def signed(r):
    return "up" if r > 0 else "down" if r < 0 else "flat"

print(list(map(signed, rets)))
```

## `filter` — keep items that pass a test

`lambda` is a tiny throwaway function written inline: `lambda r: r > 0` is "a
function that takes r and returns r > 0".

```python
print(list(filter(lambda r: r > 0, rets)))
```

## `sorted(key=...)` — the most useful one in practice

Pass a function that says *what to sort by*.

```python
holdings = [("MSFT", 12), ("BND", 40), ("CASH", 1)]
by_size = sorted(holdings, key=lambda pair: pair[1], reverse=True)
print(by_size)
```

## `reduce` — boil a list down to one value

Here we compound returns into the growth of £1.

```python
from functools import reduce

wealth = reduce(lambda w, r: w * (1 + r), rets, 1.0)
print("growth of 1:", round(wealth, 4))
```

## Returning a function

A function can build and hand back another function. This is the seed of
**decorators**, which you'll meet in the next track.

```python
def twice(fn):
    def wrapped(x):
        return fn(fn(x))
    return wrapped

add_one = lambda x: x + 1
add_two = twice(add_one)
print("apply add_one twice to 3:", add_two(3))
```
