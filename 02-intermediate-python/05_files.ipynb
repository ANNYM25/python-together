# 5 · Files — reading and writing data on disk

CSVs, logs, notes, model outputs — real data lives in files. Python's `pathlib`
makes files feel like objects, and the `with` block makes sure they're closed
properly even if something goes wrong.

> We write to a temporary folder so this runs safely in Colab or anywhere.

```python
import tempfile
from pathlib import Path

workdir = Path(tempfile.gettempdir())
sample = workdir / "sample_returns.csv"
print("working with:", sample)
```

## Writing a whole file at once

```python
sample.write_text(
    "date,ret\n"
    "2024-01-02,0.01\n"
    "2024-01-03,-0.004\n",
    encoding="utf-8",
)
print("wrote the file.")
```

## Reading it back

```python
text = sample.read_text(encoding="utf-8")
print(text)
```

## Reading line by line with a `with` block

`with open(...) as f:` guarantees the file is closed when the block ends. We
parse the header, then turn each remaining line into a dictionary.

```python
rows = []
with sample.open(encoding="utf-8") as handle:
    header = handle.readline().strip().split(",")        # ['date', 'ret']
    for line in handle:
        date, ret = line.strip().split(",")
        rows.append({header[0]: date, header[1]: float(ret)})

print(rows)
```

> In the data-science track you'll let **pandas** do all of this in one line
> (`pd.read_csv`). It's worth seeing the manual version once so pandas isn't
> magic.
