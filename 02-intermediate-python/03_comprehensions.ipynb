# 3 · Comprehensions — transform a collection in one line

You met list comprehensions briefly in the foundations track. They're so common
in Python (and in data work) that they deserve their own lesson. A comprehension
builds a new collection from an old one, often replacing a 4-line loop with one
readable line.

```python
xs = range(-2, 3)     # -2, -1, 0, 1, 2
```

## List comprehension: `[ f(x) for x in things ]`

```python
squares = [x * x for x in xs]
print("squares:", squares)
```

## Add a filter: `[ f(x) for x in things if condition ]`

```python
positives = [x for x in xs if x > 0]
print("positives only:", positives)
```

## Transform *and* choose with an inline if/else

```python
signs = ["non-neg" if x >= 0 else "neg" for x in xs]
print("signs:", signs)
```

## Dict and set comprehensions — same idea, different brackets

```python
labels = {x: ("non-neg" if x >= 0 else "neg") for x in xs}   # {} with key: value
unique_abs = {abs(x) for x in xs}                            # {} without colon = set
print("dict :", labels)
print("set  :", unique_abs)
```

## Nested: flatten a small matrix

Read it left-to-right as nested loops: "for each row, for each cell".

```python
matrix = [[1, 2], [3, 4]]
flat = [cell for row in matrix for cell in row]
print("flattened:", flat)
```

> **Taste note:** comprehensions shine when short. If one grows past a line or
> two, or nests three deep, a plain `for` loop is kinder to the next reader.

- `[f(x) for x in xs]` builds a list; add `if cond` to filter.
