# 2 · Modules, imports, and virtual environments

You don't write everything yourself. A **module** is a file of ready-made
functions you pull in with `import`. Python ships hundreds (the "standard
library"), and millions more live on PyPI (NumPy, pandas, ...).

## Importing

Three common forms:

````python
import math                    # whole module -> use math.sqrt
from pathlib import Path       # one name out of a module
import statistics as stats     # rename on import

print(math.sqrt(16))
print(stats.mean([1, 2, 3, 4]))
print(Path(".").resolve())     # where am I running?
````

## A tour of handy standard-library modules

````python
import random
from datetime import date

random.seed(0)
print("a random pick:", random.choice(["equity", "bond", "cash"]))
print("today is:", date.today())
print("pi to 3 dp:", round(math.pi, 3))
````

## Virtual environments (the one setup habit that saves you)

A **virtual environment** is an isolated copy of Python's third-party library
folder, so *this* project's NumPy can't fight *another* project's NumPy. You
make one per project. This is a terminal task, not notebook code — shown here
so you know the ritual:

```bash
python3 -m venv .venv            # create it (once)
source .venv/bin/activate        # turn it on   (Windows: .venv\Scripts\activate)
pip install -r requirements.txt  # install this lab's libraries into it
pip list                         # now shows THIS project's packages only
```

When your prompt shows `(.venv)`, you're inside it. `requirements.txt` records
what to install; never commit the `.venv/` folder itself.

> In **Colab** you skip all of this — the environment is pre-made in the cloud.
