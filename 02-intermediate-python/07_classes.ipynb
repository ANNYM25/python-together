# 7 · Classes — data and behaviour in one bundle

A dictionary holds data. A **class** holds data *and* the functions that act on
it, as one object. When a "thing" in your program starts growing its own
behaviour — a Position that can value itself, a Model that can predict — a class
keeps it tidy. (sklearn models are classes: `.fit()`, `.predict()`.)

## Defining a class

`__init__` is the setup method that runs when you create an object. `self` is
the object itself — how each instance refers to its own data.

```python
class Position:
    def __init__(self, ticker, shares, price):
        self.ticker = ticker       # store data ON the object
        self.shares = shares
        self.price = price

    def market_value(self):        # behaviour that uses the object's data
        return self.shares * self.price

lot = Position("BND", shares=40, price=72.1)
print(lot.ticker, "is worth", round(lot.market_value(), 2))
```

## A friendly `__repr__`

Add `__repr__` so printing the object shows something useful instead of
`<...object at 0x...>`. (This is a "dunder" method — more in the next track.)

```python
class Position:
    def __init__(self, ticker, shares, price):
        self.ticker, self.shares, self.price = ticker, shares, price

    def market_value(self):
        return self.shares * self.price

    def __repr__(self):
        return f"Position({self.ticker!r}, shares={self.shares}, price={self.price})"

print(Position("MSFT", 12, 410.5))
```

## A tiny model as a class — a preview of ML

This is the exact shape scikit-learn uses: build the object with settings, then
call a method to use it.

```python
class LinearModel:
    """y_hat = slope * x + intercept."""
    def __init__(self, slope, intercept=0.0):
        self.slope = slope
        self.intercept = intercept

    def predict(self, x):
        return self.slope * x + self.intercept

model = LinearModel(slope=0.8, intercept=0.1)
print("prediction at x=2:", model.predict(2.0))
```

### ✅ What you learned
- A **class** bundles data (`self.x`) and behaviour (methods) into an object.
- `__init__(self, ...)` sets the object up; `self` is the instance.
- `__repr__` controls how the object prints.
- This "build then call methods" shape is exactly how sklearn models work.
