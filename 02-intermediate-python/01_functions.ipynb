# 1 · Functions — package a formula once, reuse it forever

So far our code ran straight down the page. A **function** lets you give a
recipe a name, hand it inputs, and get an output back — so you write the logic
once and call it anywhere. Every library you'll use (NumPy, pandas, sklearn) is
really just thousands of well-named functions.

## Defining and calling

`def name(inputs):` starts a function; `return` hands a value back. Nothing
happens until you **call** it with `name(...)`.

```python
def simple_return(start, end):
    return end / start - 1

print(simple_return(100, 107))     # a stock went 100 -> 107
print(simple_return(50, 45))       # and one that fell
```

## Arguments with defaults

A parameter can have a default, so callers only pass it when they want to change
it. Here `compounds=1` means "once a year" unless you say otherwise.

```python
def future_value(principal, rate, years, compounds=1):
    """Compound interest. compounds=1 is annual, 12 is monthly."""
    return principal * (1 + rate / compounds) ** (compounds * years)

print("annual :", round(future_value(1000, 0.05, 10), 2))
print("monthly:", round(future_value(1000, 0.05, 10, compounds=12), 2))
```

That triple-quoted line under `def` is a **docstring** — a note on what the
function does. `help(future_value)` prints it.

## Returning more than one thing

Return a tuple and unpack it on the other side — handy for "value and its parts".

```python
def split_payment(total, parts):
    chunk = total // parts
    leftover = total % parts
    return chunk, leftover

each, left = split_payment(2500, 4)
print("each:", each, " leftover:", left)
```
