# 4 · Exceptions — when things go wrong on purpose

Bad input is normal: an empty field, a divide-by-zero, a file that isn't there.
Python signals these with an **exception**. Your job is to decide where to
recover gracefully and where to let the program stop loudly. The tool is
`try / except`.

## The basic shape

```python
try:
    value = float("12%")          # this fails — "12%" isn't a number
except ValueError as exc:
    print("couldn't convert:", exc)
```

Without the `try`, that line would crash the whole program. With it, we catch
the specific `ValueError` and keep going.

## Catch what you can handle; re-raise what you can't

Catch the *specific* error you expect (not a blanket `except:`). You can also
add your own clearer message and re-raise.

```python
def parse_weight(text):
    try:
        value = float(text)
    except ValueError as exc:
        raise ValueError(f"not a number: {text!r}") from exc
    if not 0 <= value <= 1:
        raise ValueError("weights must be between 0 and 1")
    return value

print("good input:", parse_weight("0.25"))
try:
    parse_weight("1.5")
except ValueError as exc:
    print("rejected:", exc)
```

## Returning a safe default instead of crashing

Sometimes "couldn't compute" should just mean "give me `None`". `finally` runs
no matter what — perfect for cleanup (closing files, stopping timers).

```python
def safe_ratio(numer, denom):
    try:
        return numer / denom
    except ZeroDivisionError:
        return None
    finally:
        pass   # e.g. close a file here

print("1/0 ->", safe_ratio(1, 0))
print("6/3 ->", safe_ratio(6, 3))
```

### ✅ What you learned
- Errors raise **exceptions**; `try/except` lets you catch and recover.
- Catch the **specific** exception type (`ValueError`, `ZeroDivisionError`, ...).

- `finally` always runs — use it for cleanup.
