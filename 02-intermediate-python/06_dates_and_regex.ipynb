# 6 · Dates and regular expressions

Two everyday chores in data work: handling **dates** (prices have timestamps),
and finding **patterns** in text with regular expressions (reports have ISINs,
tickers, percentages). Both have a small learning curve and huge payoff.

## Dates with `datetime`

```python
from datetime import datetime, timedelta

start = datetime(2024, 1, 15)
print("ISO date:", start.date().isoformat())
print("a week later:", (start + timedelta(days=7)).date())
print("days between:", (datetime(2024, 3, 1) - start).days)
```

Parse a date *out of* text with `strptime`, and format one *into* text with
`strftime`.

```python
parsed = datetime.strptime("2024-12-31", "%Y-%m-%d")
print("parsed:", parsed.date())
print("pretty:", parsed.strftime("%d %b %Y"))
```

## Regular expressions with `re`

A regex is a mini-language for describing text patterns. `re.findall` pulls out
every match. Here we extract **ISIN** codes (2 letters + 9 alphanumerics + 1 digit).

```python
import re

raw = "ISIN US0378331005 weight 0.08; ISIN US5949181045 weight 0.11"
isins = re.findall(r"[A-Z]{2}[A-Z0-9]{9}[0-9]", raw)
print("ISINs found:", isins)
```

Quick tour of the pieces you just used:

| Piece | Means |
|-------|-------|
| `[A-Z]` | any one capital letter |
| `[A-Z0-9]` | a capital letter or digit |
| `{2}` / `{9}` | exactly that many of the thing before it |
| `\d` | a digit (same as `[0-9]`) |
| `+` | one or more |

## Validate a whole string with `fullmatch`

```python
print("MSFT a valid ticker?", bool(re.fullmatch(r"[A-Z]{1,5}", "MSFT")))
print("msf7 valid?        ", bool(re.fullmatch(r"[A-Z]{1,5}", "msf7")))
```
