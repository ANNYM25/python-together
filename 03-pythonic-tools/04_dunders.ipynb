# 4 · Dunder methods — make your objects feel built-in

Ever wonder why `df + df`, `len(df)`, and `df[0]` just *work* on pandas objects?
Because those classes implement **dunder** ("double underscore") methods like
`__add__`, `__len__`, `__getitem__`. Implement them on your own class and Python
treats your object like a native one. We'll build a `Money` type.

## The setup: a Money class storing cents

`__init__` builds it, `__repr__` controls how it prints. `__slots__` is a small
memory optimisation that fixes which attributes an instance may have.

```python
class Money:
    __slots__ = ("cents",)

    def __init__(self, cents):
        self.cents = int(cents)

    def __repr__(self):
        sign = "-" if self.cents < 0 else ""
        v = abs(self.cents)
        return f"{sign}${v // 100}.{v % 100:02d}"

print(Money(1250))     # -> $12.50  thanks to __repr__
```

## `__add__` makes `+` work

Define `__add__(self, other)` and Python routes `a + b` to it.

```python
class Money:
    __slots__ = ("cents",)
    def __init__(self, cents): self.cents = int(cents)
    def __repr__(self):
        sign = "-" if self.cents < 0 else ""; v = abs(self.cents)
        return f"{sign}${v // 100}.{v % 100:02d}"

    def __add__(self, other):
        return Money(self.cents + other.cents)

    def __eq__(self, other):
        return isinstance(other, Money) and self.cents == other.cents

fee = Money(1250)
tax = Money(80)
print("fee + tax =", fee + tax)
print("equal?", Money(100) == Money(100))
```

## Common dunders worth knowing

| Dunder | Enables |
|--------|---------|
| `__init__` | construction |
| `__repr__` | how it prints |
| `__add__`, `__sub__`, `__mul__` | `+ - *` |
| `__eq__`, `__lt__` | `==`, `<` (sorting!) |
| `__len__` | `len(obj)` |
| `__getitem__` | `obj[i]` |
