# 2 · Iterators and generators — one item at a time

You can't always load everything into memory — a day of market ticks, a huge
log, a streaming dataset. A **generator** produces values lazily, one at a time,
with `yield`. It remembers where it left off and only computes the next value
when asked.

## `yield` makes a generator

A rolling (moving) average: it walks the series once and emits each window's
average as it goes, never holding more than it needs.

```python
def moving_average(series, window):
    if window <= 0:
        raise ValueError("window must be positive")
    total = 0.0
    for i, value in enumerate(series):
        total += value
        if i >= window:
            total -= series[i - window]      # drop the value leaving the window
        if i + 1 >= window:
            yield total / window

xs = [1, 2, 3, 4, 5]
print("3-day moving average:", [round(v, 2) for v in moving_average(xs, 3)])
```

## Generator expressions — like a list comprehension, but lazy

Use round brackets instead of square. It computes nothing until consumed, so
it's memory-cheap for big data.

```python
squares = (x * x for x in xs)      # note: ( ) not [ ]
print("sum of squares:", sum(squares))
```

## The one gotcha: a generator is single-use

Once you've walked it, it's empty. This surprises everyone once.

```python
squares = (x * x for x in xs)
print("first pass:", list(squares))
print("second pass:", list(squares), "  <- already exhausted!")
```
