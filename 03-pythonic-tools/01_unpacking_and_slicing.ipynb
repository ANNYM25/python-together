# 1 · Unpacking and slicing — moving data without loops

Welcome to the "pythonic tools" track — the techniques that make Python code
look effortless. First up: **unpacking** (pulling a sequence apart into named
pieces) and **slicing** (grabbing sub-sequences). Train/test splits, n-grams,
and batching are all slice problems.

## Star-unpacking: `first, *middle, last`

The `*` soaks up "everything else" into a list.

```python
elems = [1, 2, 3, 4, 5]
first, *middle, last = elems
print("first:", first, " middle:", middle, " last:", last)
```

## Slicing: `seq[start:stop:step]`

Every part is optional. A negative step walks backwards.

```python
days = list(range(10))
print("every 2nd:", days[::2])
print("reversed :", days[::-1])
print("last three:", days[-3:])
```

## Slice assignment — insert/replace without a loop

Assigning to a slice edits the list in place. `days[1:1] = [...]` inserts at
position 1 without removing anything.

```python
days[1:1] = [99]
print("after insert:", days)
```

## A real use: a lazy n-gram generator

N-grams (sliding windows of `n` tokens) power classic language models — and
this is exactly the sliding window your `10-projects/` tiny LLM uses.

```python
def ngrams(tokens, n):
    for i in range(len(tokens) - n + 1):
        yield tuple(tokens[i:i + n])

tokens = "rates stay higher for longer".split()
print(list(ngrams(tokens, 2)))
```

- `first, *rest, last = seq` unpacks with a "soak up the middle" star.
