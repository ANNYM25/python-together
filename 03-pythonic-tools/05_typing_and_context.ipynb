# 5 · Type hints and context managers

Two professional-polish tools. **Type hints** document what shapes go in and out
(and let editors catch bugs before you run). **Context managers** (`with`)
guarantee setup/teardown happens — closing files, releasing locks — even if the
code inside explodes.

## Type hints

Hints don't change how code runs — Python ignores them at runtime — but they
document intent and power autocomplete and tools like `mypy`.

```python
def sharpe(mean_excess: float, vol: float) -> float:
    """Return per unit of risk. Higher is better."""
    if vol <= 0:
        raise ValueError("vol must be positive")
    return mean_excess / vol

print("Sharpe:", round(sharpe(0.06, 0.12), 3))
```

Richer hints for collections and "maybe missing" values:

```python
from typing import Optional

def total(weights: list[float]) -> float:
    return sum(weights)

def label(x: float, note: Optional[str] = None) -> str:
    return f"{x}" + (f" ({note})" if note else "")

print(total([0.2, 0.3, 0.5]))
print(label(0.5), "|", label(0.5, "cash"))
```

## Context managers: the `with` statement

You've already used `with open(...)`. You can write your own with
`@contextmanager`: code before `yield` is setup, code after is teardown, and the
teardown runs **even if the body raises**.

```python
from contextlib import contextmanager

@contextmanager
def announce(label):
    print(f"-> start {label}")
    try:
        yield
    finally:
        print(f"<- end {label}")      # always runs

with announce("training"):
    print("   ...pretend to train a model...")
```

Prove the teardown is guaranteed even on error:

```python
try:
    with announce("risky step"):
        raise RuntimeError("boom")
except RuntimeError as e:
    print("caught:", e, "- but 'end' still printed above")
```

### ✅ What you learned
- **Type hints** (`x: float -> float`, `list[float]`, `Optional[str]`) document
  intent and catch bugs via tools; they don't affect runtime.
- **Context managers** (`with`) guarantee cleanup; `@contextmanager` + `yield`
  lets you write your own.
- `try/finally` inside guarantees teardown even on error.
