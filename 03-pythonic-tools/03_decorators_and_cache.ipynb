# 3 · Decorators and caching — wrap a function with extra powers

A **decorator** wraps a function to add behaviour — timing, logging, caching —
without touching the function's own code. You met "functions returning
functions" in the intermediate track; a decorator is just that, with the `@`
shortcut. These show up in every training loop and web framework.

## Writing a timing decorator

`timed` takes a function and returns a new function that times it, then calls
the original. `@functools.wraps` keeps the original's name and docstring.

```python
import functools
import time

def timed(fn):
    @functools.wraps(fn)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = fn(*args, **kwargs)
        print(f"{fn.__name__} took {time.perf_counter() - start:.6f}s")
        return result
    return wrapper

@timed
def slow_sum(n):
    return sum(range(n))

print("result:", slow_sum(1_000_000))
```

`@timed` above `def slow_sum` is shorthand for `slow_sum = timed(slow_sum)`.
That's the whole trick.

## Caching with `functools.lru_cache`

A built-in decorator that remembers results so repeated calls are instant.
Naive recursive Fibonacci recomputes the same values exponentially — caching
collapses it to linear.

```python
@functools.lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print("fib(30):", fib(30))
print("cache stats:", fib.cache_info())   # hits show the saved work
```
